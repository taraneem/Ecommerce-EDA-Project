### Data Preparation

- Merge / Join using Pandas

```python
df = pd.merge(left= customers , right= orders , on='customer_id' , how= 'inner' )
df
```

```python
df = pd.merge(left= df , right= sales ,on= 'order_id' , how= 'inner' )
df
```

```python
df = pd.merge(left= df , right= products ,left_on='product_id',
              right_on='product_ID', how='inner' )
df 
```

- Handling quantity_x , quantity_y

```python
df = df.rename(columns={'quantity_x' :'quantity', 'quantity_y' :'Stock'})
df
```

## Data Cleaning

```python
df.columns
```

```python
# Drop Unnecessary Columns

df.drop(['customer_id', 'home_address', 'zip_code', 'country', 'order_id', 'product_id', 'product_ID', 'description'], axis= 1, inplace= True)
```

```python
# Re-Check duplicates
df.duplicated().sum()
```

```python
df.head()
```

```python
# Convert to datetime

df['order_date'] = pd.to_datetime(df['order_date'], dayfirst= True)
df['delivery_date'] = pd.to_datetime(df['delivery_date'], dayfirst= True)
```

## Feature Engineering

```python
df.head()
```

```python
df['delivery_days']= (df['delivery_date'] - df['order_date']).dt.days
df['delivery_days']
```

```python
df['month'] = df['order_date'].dt.to_period('M').dt.to_timestamp()
df['month']
```

```python
df['age'].describe()
```

```python
# Create groups based on age range 
def age_group(x):
    if x <= 35:
        return 'Young'
    elif x <= 55:
        return 'Middle Aged'
    elif x > 55: 
        return 'Senior'
    
df['age group'] = df['age'].apply(age_group)

df['age group']
```

### Insights

- Which product category (product_type) has the highest total sales volume?

```python
plot_df = df.groupby('product_type')['sales_id'].count().sort_values(ascending=False).reset_index()
plot_df
```

```python
px.bar(data_frame= plot_df , x='product_type' , y= 'sales_id' , title='Total Orders per Product Category',
       labels={'product_type':'Product Type' , 'sales_id':'Total Orders'}, text_auto=True , color_discrete_sequence=['gray'])
```

- Which specific product (product_name) generated the most revenue?

```python
plot_df = df.groupby('product_name')['total_price'].sum().sort_values(ascending=False).reset_index().head(10)
plot_df
```

```python
px.bar(data_frame= plot_df , x='product_name' , y= 'total_price' , title='Total Revenue per Product',
       labels={'product_name':'Product Name' , 'total_price':'Total Revenue'}, text_auto=True , color_discrete_sequence=['gray'])
```

- How do sales trends look on a monthly basis?

```python
df['month'] = df['order_date'].dt.to_period('M').dt.to_timestamp()
df['month']
```

```python
df_sorted = df.sort_values(by= 'order_date')
df_sorted
```

```python
plot_df= df_sorted.groupby('month')['total_price'].sum().reset_index()
plot_df
```

```python
px.line(data_frame= plot_df , x= 'month' , y= 'total_price' , title='Total Revenue per Month' , markers= True)
```

```python
plot_df= df_sorted.groupby(['month', 'product_type'])['total_price'].sum().reset_index()
plot_df
```

```python
px.line(data_frame= plot_df , x= 'month' , y= 'total_price' , title='Total Revenue per Month' ,color= 'product_type', markers= True)
```
